## 6. Detect Training, Validation and Official Test Folders

````python
def find_split_dir(
    root,
    possible_names
):
    names = {
        x.lower()
        for x in possible_names
    }

    matches = [
        p
        for p in root.rglob("*")
        if p.is_dir()
        and p.name.lower() in names
    ]

    if not matches:
        return None

    matches = sorted(
        matches,
        key=lambda p: (
            -len([
                x
                for x in p.iterdir()
                if x.is_dir()
            ]),
            len(p.parts)
        )
    )

    return matches[0]


TRAIN_DIR = find_split_dir(
    SEARCH_ROOT,
    ["train", "training"]
)

VAL_DIR = find_split_dir(
    SEARCH_ROOT,
    ["validation", "valid", "val"]
)

TEST_DIR = find_split_dir(
    SEARCH_ROOT,
    ["test", "testing"]
)

print("Training   :", TRAIN_DIR)
print("Validation :", VAL_DIR)
print("Testing    :", TEST_DIR)

assert TRAIN_DIR is not None, (
    "Training folder was not found."
)

assert VAL_DIR is not None, (
    "Validation folder was not found."
)

assert TEST_DIR is not None, (
    "Official Test folder was not found."
)
````

## 7. Exploratory Data Analysis and Class Distribution

This section checks the number of images in every class and split before model training.

````python
IMAGE_EXTS = {
    ".jpg",
    ".jpeg",
    ".png",
    ".bmp",
    ".tif",
    ".tiff"
}


def count_images(
    root
):
    rows = []

    for class_dir in sorted(
        [
            p
            for p in root.iterdir()
            if p.is_dir()
        ]
    ):
        count = sum(
            1
            for p in class_dir.rglob("*")
            if p.is_file()
            and p.suffix.lower()
            in IMAGE_EXTS
        )

        rows.append({
            "Class": class_dir.name,
            "Count": count
        })

    return pd.DataFrame(
        rows
    )


train_counts = count_images(
    TRAIN_DIR
).rename(
    columns={
        "Count": "Training"
    }
)

val_counts = count_images(
    VAL_DIR
).rename(
    columns={
        "Count": "Validation"
    }
)

test_counts = count_images(
    TEST_DIR
).rename(
    columns={
        "Count": "Testing"
    }
)

class_counts = (
    train_counts
    .merge(
        val_counts,
        on="Class",
        how="outer"
    )
    .merge(
        test_counts,
        on="Class",
        how="outer"
    )
    .fillna(0)
)

display(
    class_counts
)

print(
    "Training total:",
    int(
        class_counts[
            "Training"
        ].sum()
    )
)

print(
    "Validation total:",
    int(
        class_counts[
            "Validation"
        ].sum()
    )
)

print(
    "Testing total:",
    int(
        class_counts[
            "Testing"
        ].sum()
    )
)

class_counts.set_index(
    "Class"
)[
    [
        "Training",
        "Validation",
        "Testing"
    ]
].plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title(
    "Brain Tumor MRI — Class Distribution"
)

plt.xlabel("Class")
plt.ylabel("Number of Images")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
````

The final Kaggle dataset is already cleaned and image-preprocessed, so **CLAHE, duplicate removal and leakage removal are not repeated inside the model notebook**.

### 9.2 Custom CNN-specific preprocessing


**Model-specific preprocessing used here:** `layers.Rescaling(1.0 / 255.0)`

The final Kaggle images arrive in the normal image range of approximately `0–255`.
The Custom CNN is trained from scratch, so the notebook explicitly converts them to `0–1`.


### 9.3 Training-only augmentation

Random flip, rotation, translation, zoom and contrast are used only while Training. Validation and Official Test images are not randomly augmented.

This separation is important: **group preprocessing prepares the dataset once; model-specific preprocessing prepares the tensor for the selected architecture.**

````python

# MODEL-SPECIFIC PREPROCESSING DEMONSTRATION
# Model: Custom CNN
# Mode : scratch


PREPROCESS_DEMO_IMAGES, PREPROCESS_DEMO_LABELS = next(
    iter(train_ds)
)

PREPROCESS_DEMO_SAMPLE = PREPROCESS_DEMO_IMAGES[:4]

print("Model:", "Custom CNN")
print("Mode :", "scratch")
print("Input shape:", PREPROCESS_DEMO_SAMPLE.shape)
print(
    "Input pixel range:",
    float(tf.reduce_min(PREPROCESS_DEMO_SAMPLE)),
    "to",
    float(tf.reduce_max(PREPROCESS_DEMO_SAMPLE))
)

PREPROCESS_DEMO_OUTPUT = layers.Rescaling(
    1.0 / 255.0
)(
    PREPROCESS_DEMO_SAMPLE
)

print(
    "After model-specific preprocessing:",
    float(tf.reduce_min(PREPROCESS_DEMO_OUTPUT)),
    "to",
    float(tf.reduce_max(PREPROCESS_DEMO_OUTPUT))
)

print("Transformation: Rescaling(1/255)")

# Show the same input MRI images used for this preprocessing demonstration.
plt.figure(figsize=(10, 5))

for i in range(min(4, PREPROCESS_DEMO_SAMPLE.shape[0])):
    plt.subplot(1, 4, i + 1)
    plt.imshow(
        PREPROCESS_DEMO_SAMPLE[i]
        .numpy()
        .astype(np.uint8)
    )

    class_index = int(
        np.argmax(
            PREPROCESS_DEMO_LABELS[i]
            .numpy()
        )
    )

    plt.title(
        CLASS_NAMES[class_index]
    )

    plt.axis("off")

plt.suptitle(
    "Final Group-Preprocessed MRI Inputs Before Model-Specific Transform"
)

plt.tight_layout()
plt.show()
````

````python
def make_augmentation(
    variant_name
):
    # Variant A = lighter augmentation.
    # Variant B = slightly stronger but still moderate augmentation.
    if variant_name == "A":
        rotation = 0.02
        translation = 0.02
        zoom = 0.03
        contrast = 0.05

    else:
        rotation = 0.04
        translation = 0.04
        zoom = 0.05
        contrast = 0.08

    return keras.Sequential(
        [
            layers.RandomFlip(
                "horizontal"
            ),

            layers.RandomRotation(
                rotation
            ),

            layers.RandomTranslation(
                translation,
                translation
            ),

            layers.RandomZoom(
                zoom
            ),

            layers.RandomContrast(
                contrast
            ),
        ],
        name=(
            f"augmentation_"
            f"{variant_name}"
        )
    )


print(
    "Variant A: lighter training-only augmentation"
)

print(
    "Variant B: moderate training-only augmentation"
)
````

## 10. Evaluation Metrics and Helper Functions

Metrics used:
- Accuracy
- Macro Precision
- Macro Recall
- Macro F1
- Per-class Precision / Recall / F1
- Confusion Matrix

**Macro F1 is the main model-selection metric.**

````python
def collect_true_labels(
    ds
):
    labels = []

    for _, y_batch in ds:
        labels.append(
            np.argmax(
                y_batch.numpy(),
                axis=1
            )
        )

    return np.concatenate(
        labels
    )


def evaluate_model(
    model,
    ds,
    y_true
):
    probabilities = (
        model.predict(
            ds,
            verbose=0
        )
    )

    predictions = (
        np.argmax(
            probabilities,
            axis=1
        )
    )

    metrics = {
        "accuracy": float(
            accuracy_score(
                y_true,
                predictions
            )
        ),

        "precision_macro": float(
            precision_score(
                y_true,
                predictions,
                average="macro",
                zero_division=0
            )
        ),

        "recall_macro": float(
            recall_score(
                y_true,
                predictions,
                average="macro",
                zero_division=0
            )
        ),

        "f1_macro": float(
            f1_score(
                y_true,
                predictions,
                average="macro",
                zero_division=0
            )
        ),
    }

    return (
        metrics,
        predictions,
        probabilities
    )


Y_VAL_TRUE = collect_true_labels(
    val_ds
)

Y_TEST_TRUE = collect_true_labels(
    test_ds
)


def new_macro_f1():
    return tf.keras.metrics.F1Score(
        average="macro",
        name="f1_macro"
    )
````

## 11. Manual Hyperparameter Tuning Plan — Variant A vs Variant B

Two varieties are trained to satisfy parameter-tuning and model-variety requirements.

- **Variant A — Baseline CNN:** Dense=128, Dropout=0.3, Initial LR=0.001, Fine-tune last=N/A, Fine-tune LR=N/A, Glioma weight=1.0
- **Variant B — Deeper Regularized CNN:** Dense=256, Dropout=0.4, Initial LR=0.0003, Fine-tune last=N/A, Fine-tune LR=N/A, Glioma weight=1.0

The two variants also use slightly different augmentation strengths.

### Descriptive variant names

- **Variant A:** Baseline CNN
- **Variant B:** Deeper Regularized CNN

These names are used in training output, graph titles, comparison tables, CSV files and final selected-model output.


## 12. Build the Two Model Varieties

````python
MODEL_NAME = "Custom CNN"
MODEL_SLUG = "custom_cnn"
MEMBER_NO = OWNER_MEMBER_NO
PRETRAINED = False

VARIANT_NAMES = {
    "A": "Baseline CNN",
    "B": "Deeper Regularized CNN",
}

VARIANTS = {'A': {'filters': [32, 64, 128], 'dense': 128, 'dropout': 0.3, 'learning_rate': 0.001, 'batch_norm': True, 'glioma_weight': 1.0}, 'B': {'filters': [32, 64, 128, 256], 'dense': 256, 'dropout': 0.4, 'learning_rate': 0.0003, 'batch_norm': True, 'glioma_weight': 1.0}}


def build_variant(
    variant_name
):
    config = VARIANTS[
        variant_name
    ]

    inputs = keras.Input(
        shape=(
            224,
            224,
            3
        )
    )

    x = make_augmentation(
        variant_name
    )(
        inputs
    )

    x = layers.Rescaling(
        1.0 / 255.0
    )(x)

    for filters in config[
        "filters"
    ]:
        x = layers.Conv2D(
            filters,
            3,
            padding="same",
            use_bias=False
        )(x)

        if config[
            "batch_norm"
        ]:
            x = layers.BatchNormalization()(x)

        x = layers.ReLU()(x)

        x = layers.MaxPooling2D()(x)

    x = layers.GlobalAveragePooling2D()(x)

    x = layers.Dense(
        config[
            "dense"
        ],
        activation="relu"
    )(x)

    x = layers.Dropout(
        config[
            "dropout"
        ]
    )(x)

    outputs = layers.Dense(
        NUM_CLASSES,
        activation="softmax"
    )(x)

    model = keras.Model(
        inputs,
        outputs,
        name=(
            f"Custom_CNN_"
            f"Variant_{variant_name}"
        )
    )

    model.compile(
        optimizer=keras.optimizers.Adam(
            learning_rate=config[
                "learning_rate"
            ]
        ),
        loss=keras.losses.CategoricalCrossentropy(
            label_smoothing=0.05
        ),
        metrics=[
            "accuracy",
            new_macro_f1()
        ]
    )

    return (
        model,
        None,
        config
    )
````

## 13. Train Variant A Separately — Baseline CNN

Variant A is trained in its **own cell**.  
Its best checkpoint is selected only from Validation Macro F1.

````python
def make_class_weight(
    config
):
    class_weight = {
        i: 1.0
        for i in range(
            NUM_CLASSES
        )
    }

    glioma_weight = float(
        config.get(
            "glioma_weight",
            1.0
        )
    )

    if glioma_weight != 1.0:
        matches = [
            i
            for i, name
            in enumerate(
                CLASS_NAMES
            )
            if "glioma"
            in name.lower()
        ]

        if matches:
            class_weight[
                matches[0]
            ] = glioma_weight

    return class_weight


def history_to_frame(
    history,
    phase_name
):
    df = pd.DataFrame(
        history.history
    )

    df[
        "Phase"
    ] = phase_name

    return df


def train_single_variant(
    variant_name
):
    variant_display_name = VARIANT_NAMES[
        variant_name
    ]

    print(
        "\n"
        + "=" * 75
    )

    print(
        f"TRAINING {MODEL_NAME} "
        f"VARIANT {variant_name} — "
        f"{variant_display_name}"
    )

    print(
        "=" * 75
    )

    tf.keras.backend.clear_session()
    gc.collect()

    model, base_model, config = build_variant(
        variant_name
    )

    class_weight = make_class_weight(
        config
    )

    print(
        "Class weight:",
        class_weight
    )

    phase1_path = (
        MODELS_DIR
        / (
            f"member{MEMBER_NO}_"
            f"{MODEL_SLUG}_"
            f"{variant_name}_phase1.keras"
        )
    )

    phase2_path = (
        MODELS_DIR
        / (
            f"member{MEMBER_NO}_"
            f"{MODEL_SLUG}_"
            f"{variant_name}_phase2.keras"
        )
    )

    if PRETRAINED:
        phase1_callbacks = [
            keras.callbacks.ModelCheckpoint(
                phase1_path,
                monitor="val_f1_macro",
                mode="max",
                save_best_only=True,
                verbose=1
            ),

            keras.callbacks.EarlyStopping(
                monitor="val_f1_macro",
                mode="max",
                patience=5,
                restore_best_weights=True,
                verbose=1
            ),

            keras.callbacks.ReduceLROnPlateau(
                monitor="val_loss",
                mode="min",
                factor=0.5,
                patience=2,
                min_lr=1e-7,
                verbose=1
            )
        ]

        history_phase1 = model.fit(
            train_ds,
            validation_data=val_ds,
            epochs=15,
            callbacks=phase1_callbacks,
            class_weight=class_weight,
            verbose=1
        )

        phase1_best = (
            keras.models.load_model(
                phase1_path
            )
        )

        phase1_metrics, _, _ = evaluate_model(
            phase1_best,
            val_ds,
            Y_VAL_TRUE
        )

        base_model.trainable = True

        fine_tune_last = int(
            config[
                "fine_tune_last"
            ]
        )

        for layer in base_model.layers[
            :-fine_tune_last
        ]:
            layer.trainable = False

        for layer in base_model.layers[
            -fine_tune_last:
        ]:
            if isinstance(
                layer,
                layers.BatchNormalization
            ):
                layer.trainable = False
            else:
                layer.trainable = True

        model.compile(
            optimizer=keras.optimizers.Adam(
                learning_rate=config[
                    "fine_tune_lr"
                ]
            ),
            loss=keras.losses.CategoricalCrossentropy(
                label_smoothing=0.05
            ),
            metrics=[
                "accuracy",
                new_macro_f1()
            ]
        )

        phase2_callbacks = [
            keras.callbacks.ModelCheckpoint(
                phase2_path,
                monitor="val_f1_macro",
                mode="max",
                save_best_only=True,
                verbose=1
            ),

            keras.callbacks.EarlyStopping(
                monitor="val_f1_macro",
                mode="max",
                patience=6,
                restore_best_weights=True,
                verbose=1
            ),

            keras.callbacks.ReduceLROnPlateau(
                monitor="val_loss",
                mode="min",
                factor=0.5,
                patience=2,
                min_lr=1e-7,
                verbose=1
            )
        ]

        history_phase2 = model.fit(
            train_ds,
            validation_data=val_ds,
            epochs=25,
            callbacks=phase2_callbacks,
            class_weight=class_weight,
            verbose=1
        )

        phase2_best = (
            keras.models.load_model(
                phase2_path
            )
        )

        phase2_metrics, _, _ = evaluate_model(
            phase2_best,
            val_ds,
            Y_VAL_TRUE
        )

        if (
            phase2_metrics[
                "f1_macro"
            ]
            >=
            phase1_metrics[
                "f1_macro"
            ]
        ):
            chosen_path = phase2_path
            chosen_metrics = phase2_metrics
            chosen_phase = "Fine-tuned Phase 2"
        else:
            chosen_path = phase1_path
            chosen_metrics = phase1_metrics
            chosen_phase = "Frozen-backbone Phase 1"

        history_df = pd.concat(
            [
                history_to_frame(
                    history_phase1,
                    "Phase 1"
                ),
                history_to_frame(
                    history_phase2,
                    "Phase 2"
                )
            ],
            ignore_index=True
        )

    else:
        callbacks = [
            keras.callbacks.ModelCheckpoint(
                phase1_path,
                monitor="val_f1_macro",
                mode="max",
                save_best_only=True,
                verbose=1
            ),

            keras.callbacks.EarlyStopping(
                monitor="val_f1_macro",
                mode="max",
                patience=7,
                restore_best_weights=True,
                verbose=1
            ),

            keras.callbacks.ReduceLROnPlateau(
                monitor="val_loss",
                mode="min",
                factor=0.5,
                patience=3,
                min_lr=1e-6,
                verbose=1
            )
        ]

        history = model.fit(
            train_ds,
            validation_data=val_ds,
            epochs=35,
            callbacks=callbacks,
            class_weight=class_weight,
            verbose=1
        )

        chosen_path = phase1_path

        chosen_model = (
            keras.models.load_model(
                chosen_path
            )
        )

        chosen_metrics, _, _ = evaluate_model(
            chosen_model,
            val_ds,
            Y_VAL_TRUE
        )

        chosen_phase = "From Scratch"

        history_df = history_to_frame(
            history,
            "Scratch"
        )

    result = {
        "Variant": variant_name,
        "Variant Name": variant_display_name,
        "Selected Phase": chosen_phase,
        "Dense Units": config[
            "dense"
        ],
        "Dropout": config[
            "dropout"
        ],
        "Initial LR": config[
            "learning_rate"
        ],
        "Fine Tune Last": (
            config.get(
                "fine_tune_last",
                "N/A"
            )
            if PRETRAINED
            else "N/A"
        ),
        "Fine Tune LR": (
            config.get(
                "fine_tune_lr",
                "N/A"
            )
            if PRETRAINED
            else "N/A"
        ),
        "Glioma Weight": config.get(
            "glioma_weight",
            1.0
        ),
        "Validation Accuracy": chosen_metrics[
            "accuracy"
        ],
        "Validation Precision": chosen_metrics[
            "precision_macro"
        ],
        "Validation Recall": chosen_metrics[
            "recall_macro"
        ],
        "Validation F1": chosen_metrics[
            "f1_macro"
        ],
        "_checkpoint": chosen_path,
        "_history": history_df,
        "_config": config,
    }

    return result


# ------------------------------------------------------------
# TRAIN VARIANT A ONLY IN THIS CELL
# ------------------------------------------------------------

VARIANT_A_RESULT = train_single_variant(
    "A"
)

print(
    f"\nVariant A — {VARIANT_NAMES['A']} finished."
)

display(
    pd.DataFrame([
        {
            k: v
            for k, v in VARIANT_A_RESULT.items()
            if not k.startswith("_")
        }
    ])
)
````

## 18. 5-Fold Stratified Cross-Validation — Selected Variant

The selected Variant A/B configuration is evaluated using **5-fold stratified cross-validation on the Training split only**. The model is rebuilt and retrained independently in every fold. The normal held-out Validation set and the Official Test set are not included in these folds.

The output includes Fold 1–5 Accuracy, Macro Precision, Macro Recall and **Macro F1**, plus the Mean CV Macro F1 and standard deviation. The owner-specific summary is exported as `*_5fold_cv_summary.csv` for the final six-model comparison.

> **Report screenshot:** capture the 5-fold summary table and Macro F1 chart after this cell finishes.


````python

# 5-FOLD STRATIFIED CROSS-VALIDATION — SELECTED VARIANT ONLY
# IMPORTANT:
# - Uses the Training split only.
# - Official Validation and Official Test remain untouched.
# - The selected A/B configuration is rebuilt and retrained
#   separately in every fold.


from sklearn.model_selection import StratifiedKFold

CV_N_SPLITS = 5

# Collect Training image paths and integer class labels.
CV_IMAGE_EXTS = {
    ".jpg", ".jpeg", ".png", ".bmp", ".gif", ".tif", ".tiff"
}

cv_paths = []
cv_labels = []

for class_index, class_name in enumerate(CLASS_NAMES):
    class_dir = TRAIN_DIR / class_name

    class_paths = sorted([
        p
        for p in class_dir.rglob("*")
        if p.is_file() and p.suffix.lower() in CV_IMAGE_EXTS
    ])

    cv_paths.extend([
        str(p) for p in class_paths
    ])

    cv_labels.extend([
        class_index
    ] * len(class_paths))

cv_paths = np.asarray(cv_paths)
cv_labels = np.asarray(cv_labels, dtype=np.int64)

print("5-fold CV samples:", len(cv_paths))
print(
    "Class counts:",
    {
        CLASS_NAMES[i]: int(np.sum(cv_labels == i))
        for i in range(NUM_CLASSES)
    }
)

if isinstance(IMG_SIZE, (tuple, list)):
    CV_HEIGHT = int(IMG_SIZE[0])
    CV_WIDTH = int(IMG_SIZE[1])
else:
    CV_HEIGHT = int(IMG_SIZE)
    CV_WIDTH = int(IMG_SIZE)

def _cv_load_image(path, label):
    image_bytes = tf.io.read_file(path)

    image = tf.io.decode_image(
        image_bytes,
        channels=3,
        expand_animations=False
    )

    image.set_shape([None, None, 3])

    image = tf.cast(
        image,
        tf.float32
    )

    image = tf.image.resize_with_pad(
        image,
        CV_HEIGHT,
        CV_WIDTH
    )

    label = tf.one_hot(
        label,
        depth=NUM_CLASSES
    )

    return image, label


def _make_cv_dataset(
    paths,
    labels,
    training
):
    ds = tf.data.Dataset.from_tensor_slices(
        (
            paths.astype(str),
            labels.astype(np.int64)
        )
    )

    if training:
        ds = ds.shuffle(
            buffer_size=len(paths),
            seed=SEED,
            reshuffle_each_iteration=True
        )

    ds = ds.map(
        _cv_load_image,
        num_parallel_calls=tf.data.AUTOTUNE
    )

    ds = ds.batch(
        BATCH_SIZE
    )

    ds = ds.prefetch(
        tf.data.AUTOTUNE
    )

    return ds


skf = StratifiedKFold(
    n_splits=CV_N_SPLITS,
    shuffle=True,
    random_state=SEED
)

cv_rows = []

# Preserve the normal held-out Validation workflow.
_ORIGINAL_TRAIN_DS = train_ds
_ORIGINAL_VAL_DS = val_ds
_ORIGINAL_Y_VAL_TRUE = Y_VAL_TRUE.copy()

for fold_no, (fold_train_idx, fold_val_idx) in enumerate(
    skf.split(cv_paths, cv_labels),
    start=1
):
    print(
        "\n"
        + "=" * 80
    )
    print(
        f"5-FOLD CV — FOLD {fold_no}/{CV_N_SPLITS}"
    )
    print(
        "Selected variant:",
        BEST_VARIANT,
        "—",
        BEST_VARIANT_NAME
    )
    print(
        "=" * 80
    )

    train_ds = _make_cv_dataset(
        cv_paths[fold_train_idx],
        cv_labels[fold_train_idx],
        training=True
    )

    val_ds = _make_cv_dataset(
        cv_paths[fold_val_idx],
        cv_labels[fold_val_idx],
        training=False
    )

    # evaluate_model() in this notebook reads Y_VAL_TRUE.
    Y_VAL_TRUE = cv_labels[
        fold_val_idx
    ].copy()

    fold_result = train_single_variant(
        BEST_VARIANT
    )

    cv_rows.append({
        "Fold": fold_no,
        "Accuracy": float(
            fold_result[
                "Validation Accuracy"
            ]
        ),
        "Macro Precision": float(
            fold_result[
                "Validation Precision"
            ]
        ),
        "Macro Recall": float(
            fold_result[
                "Validation Recall"
            ]
        ),
        "Macro F1": float(
            fold_result[
                "Validation F1"
            ]
        ),
    })

    tf.keras.backend.clear_session()
    gc.collect()

# Restore the original normal-validation datasets.
train_ds = _ORIGINAL_TRAIN_DS
val_ds = _ORIGINAL_VAL_DS
Y_VAL_TRUE = _ORIGINAL_Y_VAL_TRUE

cv_df = pd.DataFrame(
    cv_rows
)

print(
    "\n5-FOLD CROSS-VALIDATION RESULTS"
)

display(
    cv_df
)

fold_f1 = cv_df[
    "Macro F1"
].to_numpy()

cv_summary = pd.DataFrame([
    {
        "Member": OWNER_MEMBER_NO,
        "Owner": MODEL_OWNER,
        "IT Number": OWNER_IT_NUMBER,
        "Model": MODEL_NAME,
        "Selected Variant": BEST_VARIANT,
        "Selected Variant Name": BEST_VARIANT_NAME,
        "Fold 1 F1": float(fold_f1[0]),
        "Fold 2 F1": float(fold_f1[1]),
        "Fold 3 F1": float(fold_f1[2]),
        "Fold 4 F1": float(fold_f1[3]),
        "Fold 5 F1": float(fold_f1[4]),
        "Mean CV F1": float(
            np.mean(
                fold_f1
            )
        ),
        "Std CV F1": float(
            np.std(
                fold_f1,
                ddof=1
            )
        ),
    }
])

print(
    "\n5-FOLD CROSS-VALIDATION SUMMARY"
)

display(
    cv_summary
)

plt.figure(
    figsize=(8, 4)
)

plt.bar(
    cv_df[
        "Fold"
    ].astype(str),
    cv_df[
        "Macro F1"
    ]
)

plt.axhline(
    cv_df[
        "Macro F1"
    ].mean(),
    linestyle="--",
    label=(
        "Mean Macro F1 = "
        f"{cv_df['Macro F1'].mean():.4f}"
    )
)

plt.title(
    f"{MODEL_NAME} — 5-Fold Cross-Validation Macro F1"
)

plt.xlabel("Fold")
plt.ylabel("Macro F1")
plt.ylim(0, 1)
plt.legend()
plt.tight_layout()
plt.show()

CV_FOLDS_CSV = (
    RESULTS_DIR
    / (
        f"{MODEL_NAME.lower()}_"
        "5fold_cv_folds.csv"
    )
)

CV_SUMMARY_CSV = (
    RESULTS_DIR
    / (
        f"{MODEL_NAME.lower()}_"
        "5fold_cv_summary.csv"
    )
)

cv_file_prefix = (
    OWNER_IT_NUMBER.lower()
    + "_"
    + MODEL_NAME.lower().replace(" ", "_")
)

CV_FOLDS_CSV = (
    RESULTS_DIR
    / f"{cv_file_prefix}_5fold_cv_folds.csv"
)

CV_SUMMARY_CSV = (
    RESULTS_DIR
    / f"{cv_file_prefix}_5fold_cv_summary.csv"
)

cv_df.to_csv(
    CV_FOLDS_CSV,
    index=False
)

cv_summary.to_csv(
    CV_SUMMARY_CSV,
    index=False
)

print(
    "Saved fold results:",
    CV_FOLDS_CSV
)

print(
    "Saved CV summary:",
    CV_SUMMARY_CSV
)
````

## 20. Download the Selected Keras Model, CSV Files Package

The notebook downloads the selected `.keras` model and the most important result files. It also creates one ZIP containing all related outputs.


````python
from google.colab import files

PACKAGE_DIR = (
    WORK
    / "it25101776_final_package"
)

if PACKAGE_DIR.exists():
    shutil.rmtree(
        PACKAGE_DIR
    )

PACKAGE_DIR.mkdir(
    parents=True,
    exist_ok=True
)

PACKAGE_MODEL = (
    PACKAGE_DIR
    / (
        "Member1_"
        "custom_cnn_SELECTED.keras"
    )
)

shutil.copy2(
    FINAL_MODEL_PATH,
    PACKAGE_MODEL
)

related_files = [
    MEMBER_RESULT_CSV,
    VARIANT_CSV,
    REPORT_CSV,
    CM_CSV,
    PER_CLASS_CSV,
    HISTORY_A_CSV,
    HISTORY_B_CSV,
    METADATA_PATH,
]

for src in related_files:
    shutil.copy2(
        src,
        PACKAGE_DIR
        / Path(
            src
        ).name
    )

PACKAGE_ZIP = Path(
    shutil.make_archive(
        (
            "/content/"
            "Member1_"
            "custom_cnn_FINAL_PACKAGE"
        ),
        "zip",
        root_dir=PACKAGE_DIR
    )
)

print(
    "Downloading selected .keras model..."
)

files.download(
    str(
        PACKAGE_MODEL
    )
)

print(
    "Downloading final member comparison CSV..."
)

files.download(
    str(
        MEMBER_RESULT_CSV
    )
)

print(
    "Downloading variant comparison CSV..."
)

files.download(
    str(
        VARIANT_CSV
    )
)

print(
    "Downloading metadata JSON..."
)

files.download(
    str(
        METADATA_PATH
    )
)

print(
    "Downloading complete results ZIP..."
)

files.download(
    str(
        PACKAGE_ZIP
    )
)

# Download 5-fold cross-validation outputs.
if "CV_FOLDS_CSV" in globals():
    files.download(str(CV_FOLDS_CSV))

if "CV_SUMMARY_CSV" in globals():
    files.download(str(CV_SUMMARY_CSV))
````
